```python
#BeautifulSoup
import sys
!{sys.executable} -m pip install lxml
from bs4 import BeautifulSoup
with open("scraped_data/data1.html", "r") as f:
    html_content=f.read()
soup=BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h3")
```

```python
all_h3=soup.find_all("h3")
all_countries=[]
for h3 in all_h3:
    name=h3.get_text(strip=True)
    parent=h3.find_parent()
    cls=h3.find_parent()["class"]
    population=h3.find_next("div").select("span.country-population")[0].get_text()
    population1=h3.find_next("div").select_one("span.country-population").get_text()
    
    all_countries.append([name,population])
```

```python
import pandas as pd
df=pd.DataFrame(all_countries,columns=["Name","Population"])
```
