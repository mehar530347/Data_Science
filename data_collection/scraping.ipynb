```python
i=1
win_team=[]
while i<=25:
     URL = f"https://www.scrapethissite.com/pages/forms/?page_num={i}"
     res = requests.get(URL)
     soup = BeautifulSoup(res.text,"lxml")

     table = soup.select("table.table")
     for t in table:
         for tr in t.select("tr.team"):
             win = float(tr.select_one("td.pct").get_text(strip=True))
             if win>0.5:
                 name = tr.select_one("td.name").get_text(strip=True)
                 year = tr.select_one("td.year").get_text(strip=True)
                 win_team.append([name,year,win])
     i+=1
print(win_team)
                    
                    
        
        
        
```

```python
import pandas as pd
df=pd.DataFrame(win_team,columns=["Team Name","Year","Win%"])
df
```

```python
df.to_csv("cleaned_data/data2.csv",index=False)
```
